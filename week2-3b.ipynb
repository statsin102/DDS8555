```python
# The objective is to predict the number of rings in abalone shells.  
# Model 2 (b) of 2: multiple linear regression

# A. Import the tools 
from pathlib import Path
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
import statsmodels.api as sm

#B. Load the files. 

input_dir = Path("/kaggle/input")

train_files = list(input_dir.rglob("train.csv"))
test_files = list(input_dir.rglob("test.csv"))

print("Training files found:", train_files)
print("Testing files found:", test_files)

train = pd.read_csv(train_files[0])
test = pd.read_csv(test_files[0])

print("Training shape:", train.shape)
print("Testing shape:", test.shape)
print("Training columns:", train.columns.tolist())
print("Testing columns:", test.columns.tolist())

display(train.head())
display(test.head())

# C. Visualize Target Variable Distributions

import matplotlib.pyplot as plt

# 1. Create a side-by-side subplot canvas
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Plot 1: Original Histogram of Rings ---
axes[0].hist(train['Rings'], bins=30, color='steelblue', edgecolor='black', alpha=0.7)
axes[0].set_title('Original Distribution of Rings')
axes[0].set_xlabel('Number of Rings')
axes[0].set_ylabel('Count')
axes[0].grid(True, linestyle='--', alpha=0.5)

# --- Plot 2: Transformed Histogram using log(y+1) ---
# np.log1p calculates log(x + 1) for the entire series automatically
log1p_rings = np.log1p(train['Rings'])

axes[1].hist(log1p_rings, bins=30, color='seagreen', edgecolor='black', alpha=0.7)
axes[1].set_title('Transformed Distribution: log1p(Rings)')
axes[1].set_xlabel('log1p(Rings) Value')
axes[1].set_ylabel('Count')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# D. Correlation Matrix of Predictors and log1p(Rings)

# 1. Separate your predictor features (dropping target and metadata)
X = train.drop(columns=['Rings', 'id'], errors='ignore')

# 2. Transform categorical 'Sex' column into numeric binary fields
# drop_first=True converts 'Sex' into clear numeric indicators (e.g., Sex_I, Sex_M)
X_encoded = pd.get_dummies(X, columns=['Sex'], drop_first=True, dtype=int)

# 3. Create a temporary copy of the features and add the log-transformed target
corr_data = X_encoded.copy()
corr_data['log1p(Rings)'] = np.log1p(train['Rings'])

# 4. Generate the Correlation Matrix
print("--- Correlation Matrix of Features & log1p(Rings) ---")
correlation_matrix = corr_data.corr()

# 5. Apply the Red-White-Blue diverging color gradient
styled_matrix = correlation_matrix.style.background_gradient(
    cmap='RdBu', 
    vmin=-1, 
    vmax=1
).format(precision=3)

display(styled_matrix)


# E. Scatter Plot of Shell_weight vs. log1p(Rings)

# 1. Define the variables to plot
x_variable = train['Shell weight']
y_transformed = np.log1p(train['Rings'])

# 2. Generate the Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(x_variable, y_transformed, alpha=0.4, color='darkblue', edgecolors='none')

# 3. Add clear styling and labels
plt.title('Shell weight vs. log1p(Rings)')
plt.xlabel('Shell Weight')
plt.ylabel('log1p(Rings)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()


# F. Simple Linear Regression: Shell weight vs. log1p(Rings)

# 1. Define independent variable (X) and log-transformed target (y)
X_simple = train[['Shell weight']]
y_transformed = np.log1p(train['Rings'])

# 2. Add an intercept constant (required by statsmodels)
X_simple_with_intercept = sm.add_constant(X_simple)

# 3. Fit the OLS Regression Model and print all statistics
simple_model = sm.OLS(y_transformed, X_simple_with_intercept).fit()
print("--- Regression Model Statistics Summary ---")
print(simple_model.summary())

# 4. Generate Predictions and Extract Standardized Residuals
predicted_values = simple_model.predict(X_simple_with_intercept)
influence = simple_model.get_influence()
standardized_residuals = influence.resid_studentized_internal

# 5. Plot Standardized Residuals vs. Predicted Values
plt.figure(figsize=(8, 5))
plt.scatter(predicted_values, standardized_residuals, alpha=0.4, color='steelblue', edgecolors='none')

# Add a horizontal reference line at y = 0
plt.axhline(y=0, color='red', linestyle='--', linewidth=1.5)

# Format the labels and titles
plt.title('Standardized Residuals vs. Predicted Values (Simple Model)')
plt.xlabel('Predicted Values (ŷ)')
plt.ylabel('Standardized Residuals')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()

# Step 6: Correlation Matrix of Residuals and Remaining Features

# 1. Isolate all original independent variables (dropping target and metadata)
X_all = train.drop(columns=['Rings', 'id'], errors='ignore')

# 2. Convert categorical 'Sex' column into numeric dummy variables
X_all_encoded = pd.get_dummies(X_all, columns=['Sex'], drop_first=True, dtype=int)

# 3. Create a copy of the features and add the raw residuals from the simple model
corr_residuals_data = X_all_encoded.copy()
corr_residuals_data['Model_Residuals'] = simple_model.resid

# 4. Generate the Correlation Matrix
print("\n--- Correlation Matrix: Residuals vs. All Features ---")
correlation_residuals_matrix = corr_residuals_data.corr()

# 5. Apply the Red-White-Blue diverging color gradient
styled_residuals_matrix = correlation_residuals_matrix.style.background_gradient(
    cmap='RdBu', 
    vmin=-1, 
    vmax=1
).format(precision=3)

display(styled_residuals_matrix)


# G. Multiple Regression: Sex_I and Shell weight vs. log1p(Rings)

# 1. Prepare features and dummy encode 'Sex'
X_base = train.drop(columns=['Rings', 'id'], errors='ignore')
X_all_encoded = pd.get_dummies(X_base, columns=['Sex'], drop_first=True, dtype=int)

# 2. Select the specific features for this model and add intercept constant
X_selected = X_all_encoded[['Sex_I', 'Shell weight']]
X_selected_with_intercept = sm.add_constant(X_selected)
y_transformed = np.log1p(train['Rings'])

# 3. Fit Multiple Linear Regression Model & Print Summary Statistics
multi_model = sm.OLS(y_transformed, X_selected_with_intercept).fit()
print("--- Multiple Regression Model Statistics Summary ---")
print(multi_model.summary())

# 4. Generate Predictions and Extract Standardized Residuals
predicted_values_multi = multi_model.predict(X_selected_with_intercept)
influence_multi = multi_model.get_influence()
standardized_residuals_multi = influence_multi.resid_studentized_internal

# 5. Plot Standardized Residuals vs. Predicted Values
plt.figure(figsize=(8, 5))
plt.scatter(predicted_values_multi, standardized_residuals_multi, alpha=0.4, color='steelblue', edgecolors='none')
plt.axhline(y=0, color='red', linestyle='--', linewidth=1.5)
plt.title('Standardized Residuals vs. Predicted Values (G: Sex_I & Shell weight)')
plt.xlabel('Predicted Values (ŷ)')
plt.ylabel('Standardized Residuals')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 6. Correlation Matrix: Model Residuals vs. All Features
corr_multi_data = X_all_encoded.copy()
corr_multi_data['Model_Residuals'] = multi_model.resid

print("\n--- Correlation Matrix: Residuals vs. All Features ---")
correlation_multi_matrix = corr_multi_data.corr()

styled_multi_matrix = correlation_multi_matrix.style.background_gradient(
    cmap='RdBu', 
    vmin=-1, 
    vmax=1
).format(precision=3)

display(styled_multi_matrix)

# H. calculate error meausurements

from sklearn.metrics import mean_squared_error, mean_absolute_error

# 1. Get the true target on the original scale
y_true_original = train['Rings']

# 2. Generate predictions on the log1p scale and convert back to original scale
y_pred_log1p = multi_model.predict(X_selected_with_intercept)
y_pred_original = np.expm1(y_pred_log1p)

# 3. Clip predictions at 0 to prevent mathematically impossible negative ring counts
y_pred_original_clipped = np.clip(y_pred_original, 0, None)

# 4. Calculate the required error measurements
mse_value = mean_squared_error(y_true_original, y_pred_original_clipped)
mae_value = mean_absolute_error(y_true_original, y_pred_original_clipped)

# RMSLE can be cleanly calculated directly using our log1p variables
y_pred_log1p_clipped = np.clip(y_pred_log1p, 0, None)
rmsle_value = np.sqrt(mean_squared_error(y_transformed, y_pred_log1p_clipped))

# 5. Display the evaluation results in a clean table layout
print("--- Model G Performance Metrics (Original Rings Scale) ---")
metrics_df_g = pd.DataFrame({
    'Metric': ['MSE (Mean Squared Error)', 'MAE (Mean Absolute Error)', 'RMSLE (Root Mean Squared Logarithmic Error)'],
    'Value': [mse_value, mae_value, rmsle_value]
})
display(metrics_df_g.style.format({'Value': '{:.4f}'}))

# H. Create Final Kaggle Submission Entry

# 1. Save the 'id' column from the test set for the final output matching
test_id = test['id']

# 2. Extract features and apply dummy encoding to the test set 'Sex' column
X_test_base = test.drop(columns=['id'], errors='ignore')
X_test_encoded = pd.get_dummies(X_test_base, columns=['Sex'], drop_first=True, dtype=int)

# 3. Align features to extract exactly the same two columns used in Model G
X_test_selected = X_test_encoded[['Sex_I', 'Shell weight']].copy()

# 4. Add the intercept constant column (required to match statsmodels structure)
X_test_with_intercept = sm.add_constant(X_test_selected)

# 5. Generate predictions on the log1p scale
log1p_test_predictions = multi_model.predict(X_test_with_intercept)

# 6. Convert predictions back to the original scale using expm1
test_predictions_original = np.expm1(log1p_test_predictions)

# 7. Safety clip: ensure no impossible zero or negative values exist
clean_test_predictions = np.asarray(test_predictions_original).flatten()
clean_test_predictions = np.clip(clean_test_predictions, 0.1, None)

# 8. Assemble the exact Kaggle Submission DataFrame
submission = pd.DataFrame({
    'id': test_id.values,
    'Rings': clean_test_predictions
})

# 9. Save to a CSV file in Kaggle's working directory
submission.to_csv('submission.csv', index=False)

print("Submission file 'submission.csv' successfully created!")
display(submission.head())

from IPython.display import FileLink

# This displays a direct, clickable download link to your output file
FileLink('submission.csv')
```
