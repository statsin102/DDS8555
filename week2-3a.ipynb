```python
# The objective is to predict the number of rings in abalone shells.  
# Model 1 (a) of 2: multiple linear regression

# A. Import the tools 
from pathlib import Path
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
import statsmodels.api as sm

input_dir = Path("/kaggle/input")

train_files = list(input_dir.rglob("train.csv"))
test_files = list(input_dir.rglob("test.csv"))

print("Training files found:", train_files)
print("Testing files found:", test_files)

train = pd.read_csv(train_files[0])
test = pd.read_csv(test_files[0])

print("Training shape:", train.shape)
print("Testing shape:", test.shape)
print("Training columns:", train.columns.tolist())
print("Testing columns:", test.columns.tolist())

display(train.head())
display(test.head())

# B. regression on all variables

# 1. Separate features from the target variable ('Rings')
X = train.drop(columns=['Rings', 'id'], errors='ignore')
y = train['Rings']

# 2. Handle the categorical 'Sex' column using Dummy Encoding
# drop_first=True prevents a statistical issue called multicollinearity
# dtype=int converts the True/False flags into clear 1s and 0s
X_encoded = pd.get_dummies(X, columns=['Sex'], drop_first=True, dtype=int)

# 3. Add a constant column for the intercept (required by statsmodels)
X_with_intercept = sm.add_constant(X_encoded)

# 4. Fit the Multiple Linear Regression Model
model = sm.OLS(y, X_with_intercept).fit()
print(model.summary())

# 5. Generate, style, and display the Correlation Matrix
print("\n--- Correlation Matrix of Features ---")
correlation_matrix = X_encoded.corr()

# Style the matrix with a Red-White-Blue diverging gradient
styled_matrix = correlation_matrix.style.background_gradient(
    cmap='RdBu', 
    vmin=-1, 
    vmax=1
).format(precision=3) # Limits decimals to 3 places for easier reading

display(styled_matrix)

# C. repeat after dropping Sex_M

# 1. Separate features from the target variable ('Rings')
X = train.drop(columns=['Rings', 'id'], errors='ignore')
y = train['Rings']

# 2. Handle the categorical 'Sex' column using Dummy Encoding
X_encoded = pd.get_dummies(X, columns=['Sex'], drop_first=True, dtype=int)

# 3. Drop Sex_M from the encoded features
X_encoded_no_M = X_encoded.drop(columns=['Sex_M'])

# 4. Add a constant column for the intercept (required by statsmodels)
X_with_intercept = sm.add_constant(X_encoded_no_M)

# 5. Fit the new Multiple Linear Regression Model
model_no_M = sm.OLS(y, X_with_intercept).fit()
print(model_no_M.summary())

# D. calculate error measurements. 

from sklearn.metrics import mean_squared_error, mean_absolute_error

# 1. Generate model predictions for the training set
y_pred = model_no_M.predict(X_with_intercept)

# 2. Calculate Root Mean Squared Error (RMSE)
rmse = np.sqrt(mean_squared_error(y, y_pred))

# 3. Calculate Mean Absolute Error (MAE)
mae = mean_absolute_error(y, y_pred)

# 4. Calculate Root Mean Squared Logarithmic Error (RMSLE)
# We clip predictions at 0 because log(negative number) is undefined
y_pred_clipped = np.clip(y_pred, 0, None)
rmsle = np.sqrt(mean_squared_error(np.log1p(y), np.log1p(y_pred_clipped)))

# 5. Display the Metrics in a clean layout
print("--- Training Performance Metrics ---")
metrics_df = pd.DataFrame({
    'Metric': ['RMSE (Root Mean Squared Error)', 'MAE (Mean Absolute Error)', 'RMSLE (Root Mean Squared Logarithmic Error)'],
    'Value': [rmse, mae, rmsle]
})
display(metrics_df.style.format({'Value': '{:.4f}'}))


# E. plot standardized residuals

import matplotlib.pyplot as plt 

# 1. Get predicted values (y-hat) from your fitted model
predicted_values = model_no_M.predict(X_with_intercept)

# 2. Extract the standardized residuals from the model results
# statsmodels calculates this automatically inside 'get_influence()'
influence = model_no_M.get_influence()
standardized_residuals = influence.resid_studentized_internal

# 3. Create the Residual Plot
plt.figure(figsize=(8, 5))
plt.scatter(predicted_values, standardized_residuals, alpha=0.5, color='steelblue', edgecolors='none')

# Add a horizontal reference line at y = 0
plt.axhline(y=0, color='red', linestyle='--', linewidth=1.5)

# Format the plot labels and titles
plt.title('Standardized Residuals vs. Predicted Values')
plt.xlabel('Predicted Values (ŷ)')
plt.ylabel('Standardized Residuals')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()

# 4. Extract the raw residuals from your fitted model
raw_residuals = model_no_M.resid

# 5. Set up a side-by-side subplot canvas
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Plot 1: Histogram of Residuals ---
axes[0].hist(raw_residuals, bins=30, color='steelblue', edgecolor='black', alpha=0.7, density=True)
# Add an approximate normal distribution reference curve over the histogram
xmin, xmax = axes[0].get_xlim()
x_axis = np.linspace(xmin, xmax, 100)
normal_curve = (1 / (np.std(raw_residuals) * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_axis - np.mean(raw_residuals)) / np.std(raw_residuals))**2)
axes[0].plot(x_axis, normal_curve, color='red', linestyle='--', linewidth=2, label='Normal Distribution')

axes[0].set_title('Histogram of Residuals')
axes[0].set_xlabel('Residual Error Value')
axes[0].set_ylabel('Density')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()

# --- Plot 2: Q-Q Plot of Residuals ---
# 'line=45' adds a reference diagonal line matching a perfect normal distribution
sm.qqplot(raw_residuals, line='45', fit=True, ax=axes[1])
axes[1].set_title('Normal Q-Q Plot')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# Investigation of assumptions: 
# Non-linear 
# Heteroscedasticity


# F. plot X's against y

# Set up the style and figure aesthetics
plt.figure(figsize=(10, 6))

# Loop through each predictor column in your dataset
for column in X_encoded_no_M.columns:
    plt.figure(figsize=(8, 5))
    
    # Create a scatter plot of the current variable vs. Rings
    # alpha=0.5 helps see density since many points overlap
    plt.scatter(X_encoded_no_M[column], y, alpha=0.5, color='darkblue', edgecolors='none')
    
    # Add labels and formatting
    plt.title(f'{column} vs. Rings')
    plt.xlabel(column)
    plt.ylabel('Rings')
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Show the individual plot
    plt.show()




# G. kaggle entry 

# 1. Save the 'id' column from the test set for the submission file
test_id = test['id']

# 2. Separate test features (dropping 'id' to match training setup)
X_test = test.drop(columns=['id'], errors='ignore')

# 3. Apply Dummy Encoding to the test set 'Sex' column
X_test_encoded = pd.get_dummies(X_test, columns=['Sex'], drop_first=True, dtype=int)

# 4. Align test features with training features
X_test_encoded = X_test_encoded.reindex(columns=X_encoded_no_M.columns, fill_value=0)

# 5. Add the constant column for the regression intercept
X_test_with_intercept = sm.add_constant(X_test_encoded)

# 6. Generate predictions using your fitted model
test_predictions = model_no_M.predict(X_test_with_intercept)

# 7. CRITICAL FIX: Convert predictions into a clean 1D numpy array 
# and cap any negative values at 0.1 (since an abalone cannot have 0 or negative rings)
clean_predictions = np.asarray(test_predictions).flatten()
clean_predictions = np.clip(clean_predictions, 0.1, None)

# 8. Create the submission DataFrame with EXACT column names matching the competition
submission = pd.DataFrame({
    'id': test_id.values,
    'Rings': clean_predictions
})

# 9. Save to a CSV file in Kaggle's working directory
submission.to_csv('submission.csv', index=False)

print("Submission file successfully rebuilt!")
display(submission.head())


print("Submission file successfully created!")
display(submission.head())



# 1. Check if the true 'Rings' column exists in the test dataset
if 'Rings' in test.columns:
    from sklearn.metrics import mean_squared_error, mean_absolute_error
    
    # Calculate test metrics if the column is present
    y_test_true = test['Rings']
    test_pred_clipped = np.clip(test_predictions, 0, None)
    
    test_rmse = np.sqrt(mean_squared_error(y_test_true, test_predictions))
    test_mae = mean_absolute_error(y_test_true, test_predictions)
    test_rmsle = np.sqrt(mean_squared_error(np.log1p(y_test_true), np.log1p(test_pred_clipped)))
    
    print("--- Test Performance Metrics ---")
    test_metrics_df = pd.DataFrame({
        'Metric': ['RMSE', 'MAE', 'RMSLE'],
        'Value': [test_rmse, test_mae, test_rmsle]
    })
    display(test_metrics_df)
else:
    # Handle the missing column scenario gracefully
    print("--- Evaluation Status ---")
    print("Notice: Cannot calculate test error metrics because the true 'Rings' column is hidden in the test set.")
    print("Kaggle will calculate these metrics automatically after you upload your submission.csv file.")
    
    # Display training metrics again for reference
    print("\n--- Training Reference Metrics ---")
    display(metrics_df)

from IPython.display import FileLink

# This generates a direct clickable link to your submission file
print("Click the link below to download your file directly:")
FileLink('submission.csv')
```
